### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента


A = np.array([
    [10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]
    ])

mean = np.mean(A, axis=0)
std = np.std(A, axis=0)

normalized = (A - mean) / std

print(mean)
print(normalized)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента

A = np.array([
    [1, 2, 1],
    [2, 1, -1],
    [3, -1, 2]
])

v = np.array([9, 8, 3])
x =np.linalg.solve(A, v)
check = A @ x

print("Роз'язок:", x)
print("A @ v:",check)
print("v:",v)
print("Результати збігаються:", np.allclose(check, v))
````
